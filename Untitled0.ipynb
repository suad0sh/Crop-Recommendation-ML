```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.tree import DecisionTreeClassifier

# 1. توليد بيانات واقعية بم أن الروابط الخارجية لا تعمل
np.random.seed(42)
n_per_crop = 100
crops = [
    'rice',
    'maize',
    'chickpea',
    'kidneybeans',
    'pigeonpeas',
    'mothbeans',
    'mungbean',
    'blackgram',
    'lentil',
    'pomegranate',
    'banana',
    'mango',
    'grapes',
    'watermelon',
    'muskmelon',
    'apple',
    'orange',
    'papaya',
    'coconut',
    'cotton',
    'jute',
    'coffee',
]

data_list = []
for i, crop in enumerate(crops):
  # توزيع قيم فريدة لكل محصول لتكون البيانات ذات نمط واقعي
  N = np.random.normal(loc=20 + i * 5, scale=5, size=n_per_crop)
  P = np.random.normal(loc=30 + (i % 5) * 15, scale=5, size=n_per_crop)
  K = np.random.normal(loc=25 + (i % 7) * 10, scale=4, size=n_per_crop)
  temp = np.random.normal(loc=15 + (i % 4) * 6, scale=2, size=n_per_crop)
  humidity = np.random.normal(loc=40 + (i % 6) * 8, scale=3, size=n_per_crop)
  ph = np.random.normal(loc=5.5 + (i % 3) * 0.8, scale=0.3, size=n_per_crop)
  rainfall = np.random.normal(loc=50 + i * 8, scale=10, size=n_per_crop)

  crop_df = pd.DataFrame({
      'N': N,
      'P': P,
      'K': K,
      'temperature': temp,
      'humidity': humidity,
      'ph': ph,
      'rainfall': rainfall,
      'label': crop,
  })
  data_list.append(crop_df)

df = pd.concat(data_list, ignore_index=True)

# 2. Preprocessing: Cleaning & Outlier Handling
features = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']
for col in features:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1
  df[col] = np.clip(df[col], Q1 - 1.5 * IQR, Q3 + 1.5 * IQR)

# 3. Scaling & Encoding
X = df[features]
y = df['label']

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=features)

# 4. Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_encoded, test_size=0.2, random_state=42
)

# 5. Model Evaluation
models = {
    'Random Forest': RandomForestClassifier(random_state=42),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'KNN': KNeighborsClassifier(),
}

print('=== نتائج دقة النماذج المعدلة (Model Accuracy) ===\n')
for name, model in models.items():
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)
  acc = accuracy_score(y_test, y_pred)
  print(f'{name} Accuracy: {acc * 100:.2f}%')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. رسم أهمية الخصائص (Feature Importance)
rf_model = models['Random Forest']
importances = rf_model.feature_importances_
feature_names = features

plt.figure(figsize=(10, 5))
sns.barplot(x=importances, y=feature_names, palette='viridis')
plt.title('Feature Importance in Crop Recommendation')
plt.xlabel('Importance Score')
plt.ylabel('Features')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# 2. رسم مقارنة دقة النماذج (Model Accuracy Comparison)
accuracies = [99.77, 96.59, 100.00]
model_names = ['Random Forest', 'Decision Tree', 'KNN']

plt.figure(figsize=(8, 5))
bars = plt.bar(
    model_names,
    accuracies,
    color=['#2ecc71', '#e74c3c', '#3498db'],
    width=0.5,
)
plt.title('Model Accuracy Comparison')
plt.ylabel('Accuracy (%)')
plt.ylim(90, 105)

# إضافة الأرقام فوق الأعمدة
for bar in bars:
  yval = bar.get_height()
  plt.text(
      bar.get_x() + bar.get_width() / 2,
      yval + 0.5,
      f'{yval:.2f}%',
      ha='center',
      va='bottom',
      fontweight='bold',
  )

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
